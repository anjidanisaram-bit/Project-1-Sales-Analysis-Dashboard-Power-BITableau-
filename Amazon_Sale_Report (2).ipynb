```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\ASUS\Downloads\Amazon Sale Report.csv")
```

```python
print("Dataset size:", df.shape)
print("Duplicate rows:", df.duplicated().sum())

df.head()
```

```python
df["Unnamed: 22"].value_counts(dropna=False).head(10)
```

```python
df.drop(columns=["Unnamed: 22"], inplace=True)

print("Updated dataset size:", df.shape)
```

```python
df["ship-city"] = df["ship-city"].fillna("Unknown")
df["ship-state"] = df["ship-state"].fillna("Unknown")
df["ship-country"] = df["ship-country"].fillna("Unknown")
df["promotion-ids"] = df["promotion-ids"].fillna("No Promotion")
df["fulfilled-by"] = df["fulfilled-by"].fillna("Unknown")
```

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%m-%d-%y",
    errors="coerce"
)

print("Invalid dates:", df["Date"].isna().sum())
```

```python
df.isnull().sum()
```

```python
df[df["Amount"].isna()][
    ["Order ID", "Status", "Qty", "currency", "Amount"]
].head(10)
```

```python
missing_amount = df["Amount"].isna()

df.loc[missing_amount, "Status"].value_counts()
```

```python
df.loc[df["Amount"].isna(), "Qty"].value_counts()
```

```python
df.loc[
    df["Amount"].isna() & (df["Qty"] > 0),
    ["Order ID", "Status", "Qty", "Amount", "currency"]
].head(20)
```

```python
df.loc[
    df["Courier Status"].isna(),
    ["Status", "Fulfilment", "Qty"]
].head(10)
```

```python
df["Courier Status"] = df["Courier Status"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["ship-postal-code"] = (
    df["ship-postal-code"]
    .astype("Int64")
    .astype("string")
    .fillna("Unknown")
)
```

```python
df.loc[
    df["Amount"].isna(),
    ["Status", "Qty"]
].groupby("Status").agg(
    Missing_Amount_Records=("Qty", "size"),
    Total_Quantity=("Qty", "sum")
)
```

```python
print("Dataset shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
print("Missing values:")
print(df.isnull().sum())
```

```python
# Check data types
df.info()

# Convert sales amount and quantity to numeric
df["Amount"] = pd.to_numeric(df["Amount"], errors="coerce")
df["Qty"] = pd.to_numeric(df["Qty"], errors="coerce")

# Check results
print("Total quantity:", df["Qty"].sum())
print("Missing amounts:", df["Amount"].isna().sum())
```

```python
valid_sales = df.dropna(subset=["Amount"])

total_sales = valid_sales["Amount"].sum()
average_sales = valid_sales["Amount"].mean()
total_orders = valid_sales["Order ID"].nunique()

print("Total Recorded Sales:", round(total_sales, 2))
print("Average Recorded Amount:", round(average_sales, 2))
print("Unique Orders with Recorded Amounts:", total_orders)
```

```python
category_sales = (
    valid_sales.groupby("Category")
    .agg(
        Total_Sales=("Amount", "sum"),
        Average_Sales=("Amount", "mean"),
        Order_Count=("Order ID", "nunique")
    )
    .sort_values("Total_Sales", ascending=False)
)

category_sales
```

```python
state_sales = (
    valid_sales.groupby("ship-state")
    .agg(
        Total_Sales=("Amount", "sum"),
        Order_Count=("Order ID", "nunique")
    )
    .sort_values("Total_Sales", ascending=False)
)

state_sales.head(10)
```

```python
monthly_sales = (
    valid_sales.groupby(valid_sales["Date"].dt.to_period("M"))
    .agg(
        Total_Sales=("Amount", "sum"),
        Order_Count=("Order ID", "nunique")
    )
    .reset_index()
)

monthly_sales["Date"] = monthly_sales["Date"].astype(str)

monthly_sales
```

```python
df.to_excel(
    r"C:\Users\ASUS\Downloads\Amazon_Sales_Cleaned.xlsx",
    index=False
)

print("Cleaned Excel file saved successfully!")
```
